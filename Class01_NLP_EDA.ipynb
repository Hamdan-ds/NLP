## Removing Html Tag:

```python
def stript(data):
    p = re.compile(r'<.*?>')
    return p.sub('', data)


 # Create a regex pattern that matches any text between < and >
    # r'<.*>' means:
    #   - '<' starts the tag
    #   - '.*' matches any characters (greedy, so it grabs as much as possible)
    #   - '>' ends the tag
    # Replace all matches of the pattern with an empty string
    # This effectively removes HTML tags from the input text
```

```python
stript(text)
```

## Removing Links & Url from texts

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

## Removing Punctuation

```python
import string
string.punctuation
```

```python
pun = "Wait—did you hear that? Oh no! I can’t believe it… Well, maybe it’s fine; after all, things happen, right? Still, I wonder: should we go, stay, or—perhaps—just wait? Honestly, I’m not sure…"
```

```python
exclude = string.punctuation
```

```python
def remove_pun(text):
    for char in exclude:
        text = pun.replace(char,'')
        return text

remove_pun(pun)
```

```python
text = "Wait—did you hear that? Oh no I can’t believe it… Well, maybe it’s fine; after all, things happen, right? Still, I wonder: should we go, stay, or—perhaps—just wait? Honestly, I’m not sure…"

# Remove punctuation using str.translate
clean_text = text.translate(str.maketrans('', '', exclude))

print(clean_text)
```

```python
clean_text = re.sub(r"[^\w\s]", "", text)
print(clean_text)
```

## Stop Words  Removing

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = []

    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return" ".join(x)
remove_stopwords(clean_text)
```

## Stemmer

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = "walk walks walking walked"
stem_words(sample)
```

### Lemmatization

```python
#Lemmatization, unlike stemming, reduces the inflected words properly ensuring that the root word belongs to the language
import nltk
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer =  WordNetLemmatizer()

sentence = """Begin with the Hugging Face free course + IMDB sentiment classification project. 
This will bridge your preprocessing knowledge with transformer fine‑tuning and put you firmly on the NLP specialization path.
Would you like me to draft a personalized learning plan (4–6 weeks) """

punctuation = "?:!.,"
sentence_words = nltk.word_tokenize(sentence)

for word in sentence_words:
    if word in punctuation:
        sentence_words.remove(word)


sentence_words
print("{0:20}{1:20}".format("word","Lemma"))
for word in sentence_words:
    print("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word, pos='v')))
```
