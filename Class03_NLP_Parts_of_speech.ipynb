```python
import numpy as np
import pandas as pd
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(u"I will google about facebook")
doc.text
```

```python
doc[0]
```

```python
doc[0].pos_
```

```python
doc[0].tag_
```

```python
spacy.explain('PRP')
```

```python
doc2 = nlp(u'i will be a good father and i will achive my goals very soon')
for word in doc2:
    print(word.text,"---------------",word.pos_,word.tag_,"------------------------------",spacy.explain(word.tag_))
```
